```python
import matplotlib.pyplot as plt
import numpy as np

p = 4
q = np.arange(4, 10)
l = 1

zeta = np.full(q.shape, np.nan, dtype = float)
kappa = np.full(q.shape, np.nan, dtype=float)

zeta = (2.0 / l) * np.arccosh(np.cos(np.pi / p) / np.sin(np.pi / q))
kappa = -(zeta **2 )

plt.figure(figsize=(7, 4))
plt.plot(q, kappa, 'o-', ms=3)
plt.xlabel('q')
plt.ylabel(r'$\kappa$')
plt.title(r'$\kappa(q)$ for fixed $p, l$')
plt.grid(True, alpha=0.3)
plt.show()
```
